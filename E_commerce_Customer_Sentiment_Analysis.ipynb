```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df=pd.read_csv('/content/ecommerce-product-reviews.csv')
```

```python
df.head()
df.info()
```

```python
df.head()
```

```python
from textblob import TextBlob
df['review']=df['review'].astype(str)
df['polarity_score']=df['review'].apply(lambda x:TextBlob(x).sentiment.polarity)
def categorize(score):
  if score>0.0:
    return 'Positive'
  elif score<0.0:
    return 'Negative'
  else:
    return 'Neutral'
df['sentiment analysis']=df['polarity_score'].apply(categorize)
print(df['sentiment analysis'].value_counts())
```

```python
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
stop=stopwords.words('english')
df['keywords']=df['review'].apply(lambda x:" ".join(x.lower() for x in x.split() if x not in stop and x.isalpha()))
df.head()
```
